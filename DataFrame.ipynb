```python
import numpy as np
import pandas as pd
df=pd.DataFrame({'A':[1,2,np.nan],'B':[5,np.nan,np.nan],'C':[1,2,3]})
df
```

```python
df.columns
```

```python
df.isnull()
```

```python
df.dropna(axis=1)
```

```python
df
```

```python
#if we don't give any axis then conventionally it'll remove rows i.e axis=0

df.dropna(thresh=1)
```

```python
df.fillna(value=5)
```

```python
df['A'].fillna(value=df['A'].mean())
```

```python
df['B'].fillna(value=df['B'].mean())
```

```python
df['B'].fillna(value=df['C'].mean())
```
