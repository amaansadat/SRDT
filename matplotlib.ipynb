```python
import pandas as pd

# Categorical sample data
data = {
    'day': ['Thur', 'Fri', 'Sat', 'Sun', 'Thur', 'Fri', 'Sat', 'Sun'],
    'total_bill': [20.5, 30.1, 25.3, 22.7, 19.3, 28.4, 24.9, 27.2],
    'tip': [3.5, 4.1, 3.3, 3.7, 4.0, 3.4, 3.8, 3.2],
    'sex': ['Male', 'Female', 'Male', 'Female', 'Male', 'Female', 'Male', 'Female']
}

df = pd.DataFrame(data)

print(df)
```

```python
#Plotting Cateogorical Data
category_counts = df['sex'].value_counts()
print(category_counts)

plt.figure(figsize=(3,2))
plt.bar(category_counts.index, category_counts.values, color=['yellow', 'blue'])
plt.xlabel('Sex')
plt.ylabel('Count')
plt.title('Category-data count of sex')
plt.show()
```

```python
x= df['total_bill']
y= df['tip']
m,b = np.polyfit(x,y,1)
plt.figure(figsize=(6,4))
plt.scatter(x,y,color='blue',alpha=0.2)
plt.plot(x,m*x+b,color='red', linestyle='-', label='Best fit line')
plt.title('Total Bill vs Tip')
plt.xlabel('Total Bill')
plt.ylabel('Tip')
plt.legend()
plt.show()
```
